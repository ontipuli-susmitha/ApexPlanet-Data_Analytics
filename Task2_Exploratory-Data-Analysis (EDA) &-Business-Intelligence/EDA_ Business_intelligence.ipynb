```python
import matplotlib.pyplot as plt

df.groupby("Category")["Total_Sales"].sum().plot(kind="bar")

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
df.groupby("Gender")["Total_Sales"].sum()
```

```python
df.groupby("Gender")["Total_Sales"].sum().plot(kind="pie", autopct='%1.1f%%')

plt.title("Gender-wise Sales")
plt.ylabel("")
plt.show()
```

```python
df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False).head(5)
```

```python
df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False).head(5).plot(kind="bar")

plt.title("Top 5 Cities by Sales")
plt.ylabel("Total Sales")
plt.show()
```
